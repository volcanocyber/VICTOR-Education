## Setup

Consider the three silicate melts (below) at 0.2 GPa pressure and temperatures of 1250 °C – basalt, 
1000 °C – andesite, and 900 °C – rhyolite.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#setup for datframe with given information
oxides = ['SiO2','TiO2','Al2O3','Fe2O3','FeO','MnO','MgO','CaO','Na2O','K2O','H2O','CO2']
basalt_wt_pct = [48.77,1.15,15.90,1.33,8.62,0.17,9.67,11.16,2.43,0.08,0.03,0.01] #list of oxide weight percents in a basalt melt
andesite_wt_pct = [59.89,0.95,17.07,3.31,3.00,0.12,3.25, 5.67,3.95,2.47,0.10,0.00] #list of oxide weight percents in an andesite melt
rhyolite_wt_pct = [70.87,0.10,14.78,2.69,0.00,0.06,0.10, 0.34,6.47,4.21,0.33,0.00] #list of oxide weight percents in a rhyolite melt
molar_mass_g_per_mol = np.array([60.08, 79.88, 101.96, 159.69, 71.85, 70.94, 40.30, 56.08, 61.99, 94.20, 18.02, 44.01]) #list of molar masses for each oxide

#given temperature (degrees Celsius) and pressure (GPa) values
pressure = 0.2 
basalt_temp = 1250 
andesite_temp = 1000
rhyolite_temp = 900

#create dataframe with this information
melt_density_df = pd.DataFrame({ 
    'basalt weight percent': basalt_wt_pct,
    'andesite weight percent': andesite_wt_pct,
    'rhyolite weight percent': rhyolite_wt_pct,
    'molar mass (g/mol)': molar_mass_g_per_mol,
}, index=pd.Index(oxides, name='oxide'))

#show dataframe
melt_density_df
```

### Download DEM

Below, a simple script will run, allowing the user to easily download a DEM.

Simply specifying the bounds of the DEM, as well as the output format and source from which to pull it will result in a fresh DEM in your current directory.

The Open Topography datasets available are listed below. You will specifiy one of these later.
  -  SRTMGL3 (SRTM GL3 90m)
  -  SRTMGL1 (SRTM GL1 30m)
  -  SRTMGL1_E (SRTM GL1 Ellipsoidal 30m)
  -  AW3D30 (ALOS World 3D 30m)
  -  AW3D30_E (ALOS World 3D Ellipsoidal, 30m)
  -  SRTM15Plus (Global Bathymetry SRTM15+ V2.1)
  -  NASADEM (NASADEM Global DEM)
  -  COP30 (Copernicus Global DSM 30m)
  -  COP90 (Copernicus Global DSM 90m)
  -  EU_DTM (DTM 30m)

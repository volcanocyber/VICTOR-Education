# Chapter 3: Numerical Solution of Ordinary Differential Equations: The Euler Methods

This notebook contains the Python code examples for Chapter 3 of "Principles of Numerical Modelling in Geosciences".

In the preceding chapter, we established the fundamental role of differential equations and equipped ourselves with essential Python tools. We saw how, for simple ODEs, exact analytical solutions can be derived.

However, most real-world systems are too complex for analytical treatment. This chapter marks our first foray into developing and implementing **numerical methods** to find approximate solutions. We will begin with one of the simplest yet most foundational techniques: the **Euler method**.

## 3.1 Numerical Solutions for Ordinary Differential Equations

Numerical methods do not aim to find an exact formula for the solution. Instead, they provide a way to *approximate* the solution at a series of discrete time points. The core idea is to start from a known initial condition and then "march" forward, step by step, calculating an approximate value for the solution at each new point.

### 3.1.1 Numerical Solution of Radioactive Decay: The Explicit Euler Method

Let's return to the ODE for radioactive decay:
$$ \frac{dN}{dt} = -\lambda N(t) $$
with an initial condition $N(0) = N_0$. We replace the derivative with a **forward difference** approximation:
$$ \frac{dN}{dt} \bigg|_{t_k} \approx \frac{N^{k+1} - N^k}{\Delta t} $$
In the **Explicit Euler method**, the right-hand side is evaluated at the current, known time step $k$. Substituting this into the ODE gives:
$$ \frac{N^{k+1} - N^k}{\Delta t} = -\lambda N^k $$
Rearranging gives us the final update formula:
$$ N^{k+1} = N^k (1 - \lambda \Delta t) $$

#### Python Implementation of the Explicit Euler Method

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for the decay process and numerical solution
lambda_decay_const = 0.1  # Decay constant (lambda)
N0_initial = 100         # Initial number of atoms
T_final_time = 50        # Total simulation time
dt_time_step = 1.0         # Time step size

# Create an array to store time points
# We add dt_time_step to the end point to ensure T_final_time is included
time_points = np.arange(0, T_final_time + dt_time_step, dt_time_step)
num_steps = len(time_points)

# Initialize an array to store the numerical solution N at each time point
N_euler_solution = np.zeros(num_steps)
N_euler_solution[0] = N0_initial  # Set the initial condition

# Time integration loop using the Explicit Euler formula
for k in range(num_steps - 1): # Loop from 0 to num_steps-2
    N_euler_solution[k+1] = N_euler_solution[k] * (1 - lambda_decay_const * dt_time_step)

# Plotting the numerical solution
plt.figure(figsize=(8, 5))

# Create the label string without using an f-string
label_text = 'Euler (dt=' + str(dt_time_step) + ')'
plt.plot(time_points, N_euler_solution, 'o--', label=label_text, color='blue')

plt.xlabel("Time (arbitrary units)")
plt.ylabel("N(t) - Number of Atoms")
plt.title("Radioactive Decay: Explicit Euler Numerical Solution")
plt.legend()
plt.grid(True)
plt.show()
```

### 3.1.2 Numerical Stability of the Explicit Euler Method

The Explicit Euler method is not always stable. Its stability depends on the choice of the time step $\Delta t$.

#### Experimenting with a Larger Time Step

Let's see what happens if we use a much larger time step, $\Delta t = 15.0$, for the same problem. According to our stability analysis, this should be unstable, as $1 - \lambda \Delta t = 1 - 0.1 \times 15.0 = -0.5$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
lambda_decay_const = 0.1
N0_initial = 100
T_final_time = 50
dt_large = 15.0  # Significantly larger time step

time_points_large_dt = np.arange(0, T_final_time + dt_large, dt_large)
num_steps_large_dt = len(time_points_large_dt)

N_euler_large_dt = np.zeros(num_steps_large_dt)
N_euler_large_dt[0] = N0_initial

for k in range(num_steps_large_dt - 1):
    N_euler_large_dt[k+1] = N_euler_large_dt[k] * \
                            (1 - lambda_decay_const * dt_large)

# Plotting the result
plt.figure(figsize=(8, 5))

# Create label and title strings without f-strings
label_text = 'Euler (dt=' + str(dt_large) + ')'
title_text = "Radioactive Decay: Explicit Euler with Large $\Delta t = " + str(dt_large) + "$"

plt.plot(time_points_large_dt, N_euler_large_dt, 'o--',
         label=label_text, color='orange')
plt.axhline(0, color='black', lw=0.5, linestyle='--') # Zero line for reference
plt.xlabel("Time (arbitrary units)")
plt.ylabel("N(t) - Number of Atoms")
plt.title(title_text)
plt.legend()
plt.grid(True)
plt.show()
```

### 3.1.3 The Implicit Euler (Backward Euler) Method

An alternative is the **Implicit Euler method**, which evaluates the right-hand side of the ODE at the *next* (unknown) time level $k+1$:
$$ \frac{N^{k+1} - N^k}{\Delta t} = -\lambda N^{k+1} $$
For this linear problem, we can algebraically solve for $N^{k+1}$ to get the update formula:
$$ N^{k+1} = \frac{N^k}{1 + \lambda \Delta t} $$
This method is **unconditionally stable** for the decay problem.

#### Python Implementation of the Implicit Euler Method

Let's implement this method using the same large, unstable time step to demonstrate its stability.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
lambda_val_imp = 0.1 # Decay constant
dt_implicit = 15.0   # Time step (same as unstable explicit case)
T_final_imp = 60     # Total time
N0_imp = 100         # Initial quantity

times_imp = np.arange(0, T_final_imp + dt_implicit, dt_implicit)
num_steps_imp = len(times_imp)

N_implicit_solution = np.zeros(num_steps_imp)
N_implicit_solution[0] = N0_imp

# Time integration loop using the Implicit Euler formula
for k in range(num_steps_imp - 1):
    N_implicit_solution[k+1] = N_implicit_solution[k] / \
                               (1 + lambda_val_imp * dt_implicit)

# Plotting the implicit solution
plt.figure(figsize=(8, 5))

# Create label and title strings without f-strings
label_text = 'Implicit Euler (dt=' + str(dt_implicit) + ')'
title_text = "Radioactive Decay: Implicit Euler Solution (dt=" + str(dt_implicit) + ")"

plt.plot(times_imp, N_implicit_solution, 'o-',
         label=label_text, color='green')
plt.xlabel("Time (arbitrary units)")
plt.ylabel("N(t) - Number of Atoms")
plt.title(title_text)
plt.legend()
plt.grid(True)
plt.show()
```

### 3.1.4 Assessing Numerical Solutions: Accuracy and Error

Stability is essential, but it is not the only desirable property. We also need to assess a scheme's **accuracy**: how close the numerical solution is to the true, exact solution.

#### Visualizing Numerical vs. Analytical Solutions

A primary way to assess accuracy is to plot the numerical solution alongside the exact analytical solution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
lambda_coeff = 0.5  # A slightly larger lambda to see effects more quickly
N0_val = 100
t_max_comp = 10.0
dtComp = 0.5     # Time step for comparison

time_vals_comp = np.arange(0, t_max_comp + dtComp, dtComp)
num_steps_comp = len(time_vals_comp)

# Analytical solution
N_analytical = N0_val * np.exp(-lambda_coeff * time_vals_comp)

# Explicit Euler numerical solution
N_explicit_comp = np.zeros(num_steps_comp)
N_explicit_comp[0] = N0_val
for k in range(num_steps_comp - 1):
    N_explicit_comp[k+1] = N_explicit_comp[k] * \
                           (1 - lambda_coeff * dtComp)

# Plotting both solutions
plt.figure(figsize=(9, 6))
plt.plot(time_vals_comp, N_analytical, 'b-', lw=2, label='Analytical Solution')

# Create label and title strings without f-strings
label_text_numeric = 'Explicit Euler (dt=' + str(dtComp) + ')'
title_text = "Radioactive Decay: Analytical vs. Explicit Euler (lambda=" + str(lambda_coeff) + ")"

plt.plot(time_vals_comp, N_explicit_comp, 'ro--', markersize=5,
         label=label_text_numeric)
plt.xlabel("Time")
plt.ylabel("N(t) - Number of Atoms")
plt.title(title_text)
plt.legend()
plt.grid(True)
plt.show()
```

#### Plotting the Absolute Error

```python
# Assuming N_analytical and N_explicit_comp are available from the previous code cell
# and were computed on the same time_vals_comp grid.

# Compute absolute error
absolute_error_euler = np.abs(N_analytical - N_explicit_comp)

# Plotting the absolute error
plt.figure(figsize=(9, 6))

label_text = 'Absolute Error (Euler, dt=' + str(dtComp) + ')'
title_text = "Absolute Error of Explicit Euler for Radioactive Decay (lambda=" + str(lambda_coeff) + ", dt=" + str(dtComp) + ")"

plt.plot(time_vals_comp, absolute_error_euler, 'g-o', markersize=5,
         label=label_text)
plt.xlabel("Time")
plt.ylabel("Absolute Error |N_exact - N_numerical|")
plt.title(title_text)
plt.legend()
plt.grid(True)
plt.show()
```

#### Relative Error

```python
# Assuming N_analytical and N_explicit_comp are available from previous cells

# Compute relative error, avoiding division by zero
epsilon = 1e-15 # A small number to prevent division by zero if N_analytical is truly zero
relative_error_euler = np.abs(N_analytical - N_explicit_comp) / (np.abs(N_analytical) + epsilon)

# Plotting the relative error
plt.figure(figsize=(9, 6))

label_text = 'Relative Error (Euler, dt=' + str(dtComp) + ')'
title_text = "Relative Error of Explicit Euler for Radioactive Decay (lambda=" + str(lambda_coeff) + ", dt=" + str(dtComp) + ")"

plt.plot(time_vals_comp, relative_error_euler, 'm-s', markersize=5,
         label=label_text)
plt.xlabel("Time")
plt.ylabel("Relative Error")
plt.title(title_text)
plt.legend()
plt.grid(True)
plt.show()
```

## 3.2 Challenges with Nonlinear ODEs: Implicit Methods and Root-Finding

When an implicit method is applied to a **nonlinear ODE**, the resulting algebraic equation for the unknown future value is also nonlinear and cannot be solved by simple rearrangement. This requires the use of a **root-finding method**.

### 3.2.1 Example: Water Infiltration into Soil – A Nonlinear ODE

We consider a simplified model for water drainage from a soil column, where the rate of change of volumetric water content $\theta$ is a nonlinear function of $\theta$ itself:
$$ \frac{d\theta}{dt} = -K_s \theta^m $$
where $m > 1$ is a nonlinearity parameter.

#### Explicit Euler Scheme for the Infiltration ODE

As requested, here is a full, runnable implementation of the explicit Euler method for this problem, including plotting.

```python
# Full implementation of the Explicit Euler method for the nonlinear infiltration ODE
import numpy as np
import matplotlib.pyplot as plt

# Simulation Parameters
Ks = 1e-5          # Saturated hydraulic conductivity
m_exponent = 3     # Nonlinearity exponent
dt = 10000.0       # Time step (s)
Tmax = 500000.0    # Total time
theta0 = 0.4       # Initial water content

# Time array and solution initialization
t_array = np.arange(0, Tmax + dt, dt)
theta_explicit_sol = np.zeros_like(t_array)
theta_explicit_sol[0] = theta0

# Time integration loop (core part)
for k_step in range(len(t_array) - 1):
    theta_prev = theta_explicit_sol[k_step]
    theta_explicit_sol[k_step+1] = theta_prev - \
                                   dt * Ks * (theta_prev ** m_exponent)

# Plotting the result
plt.figure(figsize=(9,6))
plt.plot(t_array, theta_explicit_sol, 'o-',
         label='Explicit Euler Solution')
plt.xlabel("Time (s)")
plt.ylabel("Volumetric Water Content $\\theta$")
plt.title("Nonlinear Soil Drainage (Explicit Euler)")
plt.legend()
plt.grid(True)
plt.show()
```

#### Implicit Euler Scheme for the Infiltration ODE: The Challenge

Applying the Implicit Euler method to this nonlinear ODE results in the following nonlinear algebraic equation for the unknown $\theta^{k+1}$:
$$ \theta^{k+1} + \Delta t \cdot K_s \cdot (\theta^{k+1})^m - \theta^k = 0 $$
To solve this, we need a root-finding method to find the value $X = \theta^{k+1}$ that makes the function $G(X) = X + \Delta t K_s X^m - \theta^k$ equal to zero.

### 3.2.2 Root-Finding: The Bisection Method

#### A Simple Bisection Example in Python

To learn the bisection method, let's first apply it to find a root of a simple quadratic function, $f(x) = x^2 - 4x + 3$. Note that for the formatted table output, this script uses f-strings, a modern Python feature for string formatting that will be formally introduced in Chapter 4.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function whose root we want to find
def example_function(x):
    return x**2 - 4*x + 3

# Bisection method parameters
a_init = -0.5  # Left endpoint of the initial interval
b_init = 1.5  # Right endpoint of the initial interval
tolerance = 1e-5 # Desired tolerance for the root
max_iterations = 20 # Maximum number of iterations

# Check if the initial interval brackets a root
f_a_init = example_function(a_init)
f_b_init = example_function(b_init)

if f_a_init * f_b_init >= 0:
    print("Initial interval does not bracket a root, or a root is at an endpoint.")
    # Handle this case as needed

# Variables for the bisection algorithm
a = a_init
b = b_init
iterations = 0
midpoints_history = [] # To store midpoints for plotting

print(f"Bisection for f(x) = x^2 - 4x + 3, interval [{a:.2f}, {b:.2f}]")
print("Iter |    a    |    b    |    c    |   f(c)   | Interval Width")
print("-" * 60)

for k_iter in range(max_iterations):
    iterations += 1
    c = (a + b) / 2.0
    f_c = example_function(c)
    midpoints_history.append(c)
    interval_width = b - a

    print(f"{iterations:4d} | {a:7.4f} | {b:7.4f} | {c:7.4f} | {f_c:8.2e} | {interval_width:14.4e}")

    if abs(f_c) < tolerance or interval_width / 2.0 < tolerance:
        print(f"\nConverged to root: {c:.5f} after {iterations} iterations.")
        break

    # Update the interval
    if example_function(a) * f_c < 0: # Root is in [a, c]
        b = c
    else: # Root is in [c, b]
        a = c
else: # Executed if the loop finishes without a 'break'
    print(f"\nMax iterations ({max_iterations}) reached. Best estimate: {c:.5f}")

# Plotting the function and the bisection steps
x_plot = np.linspace(min(a_init, b_init) - 0.5, max(a_init, b_init) + 0.5, 200)
y_plot = example_function(x_plot)

plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_plot, label='$f(x) = x^2 - 4x + 3$')
plt.axhline(0, color='black', lw=0.5) # x-axis

# Plot initial interval points
plt.plot([a_init, b_init], [example_function(a_init), example_function(b_init)],
         'ko', markersize=8, label='Initial Interval [a,b]')

# Plot midpoints from iterations
midpoints_array = np.array(midpoints_history)
plt.plot(midpoints_array, example_function(midpoints_array),
         'r.', markersize=10, label='Midpoints (c_k)')
for i, xc in enumerate(midpoints_array):
    plt.text(xc, example_function(xc) + 0.2, f'$m_{i+1}$', color='red', fontsize=9)


plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Bisection Method Example')
plt.legend()
plt.grid(True)
plt.show()
```

#### Implementing the Implicit Infiltration Solver with Bisection

Now we combine these ideas to solve the nonlinear soil water drainage problem using the Implicit Euler method, with the bisection algorithm embedded in each time step to solve for the unknown $\theta^{k+1}$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function G(X) whose root we need to find at each time step
def G_infiltration(X_current, theta_prev, dt, Ks, m_exponent):
    """
    Calculates G(X) = X + dt*Ks*X^m - theta_prev for root finding.
    """
    term_nonlinear = dt * Ks * (X_current ** m_exponent)
    return X_current + term_nonlinear - theta_prev

# Simulation Parameters
Ks_val = 1e-5          # Saturated hydraulic conductivity
m_val = 3              # Nonlinearity exponent
dt_val = 10000.0       # Time step (s) - can be larger for implicit
Tmax_val = 500000.0    # Total time
theta0_val = 0.4       # Initial water content

tol_bisect = 1e-7      # Tolerance for bisection
max_iter_bisect = 100  # Max iterations for bisection

# Time array and solution initialization
t_points_impl = np.arange(0, Tmax_val + dt_val, dt_val)
theta_implicit_sol = np.zeros_like(t_points_impl)
theta_implicit_sol[0] = theta0_val

# --- Time integration (Implicit Euler with Bisection Inlined) ---
for k_idx in range(len(t_points_impl) - 1):
    theta_k_prev = theta_implicit_sol[k_idx] # This is theta^k

    if theta_k_prev < tol_bisect: # If effectively zero, next step is also zero
        theta_implicit_sol[k_idx+1] = 0.0
        continue

    # Bisection method to find X = theta^{k+1}
    a_interval = 0.0
    b_interval = theta_k_prev

    Ga = G_infiltration(a_interval, theta_k_prev, dt_val, Ks_val, m_val)

    root_found_in_step = False
    for iter_num in range(max_iter_bisect):
        c_midpoint = (a_interval + b_interval) / 2.0

        if (b_interval - a_interval) / 2.0 < tol_bisect: # Interval width check
            theta_implicit_sol[k_idx+1] = c_midpoint
            root_found_in_step = True
            break

        Gc = G_infiltration(c_midpoint, theta_k_prev, dt_val, Ks_val, m_val)

        if abs(Gc) < tol_bisect: # Function value check
            theta_implicit_sol[k_idx+1] = c_midpoint
            root_found_in_step = True
            break

        if Ga * Gc < 0: # Root is in [a, c]
            b_interval = c_midpoint
        else: # Root is in [c, b]
            a_interval = c_midpoint
            Ga = Gc # Update Ga for the new 'a'

    if not root_found_in_step: # If max_iter reached
        theta_implicit_sol[k_idx+1] = (a_interval + b_interval) / 2.0

# --- Plotting ---
plt.figure(figsize=(9,6))

label_text = 'Implicit Euler (Bisection, dt=' + str(dt_val) + ')'

plt.plot(t_points_impl, theta_implicit_sol, 'o-',
         label=label_text)
plt.xlabel("Time (s)")
plt.ylabel("Volumetric Water Content $\\theta$")
plt.title("Nonlinear Soil Water Drainage (Implicit Euler with Bisection)")
plt.legend()
plt.grid(True)
plt.show()
```

### E3.1: Explicit Euler for Isotope Geochemistry - Sr Evolution in Magma

In a simplified model of magma mixing or crustal assimilation, the isotopic ratio of an element can change over time. Consider the evolution of the $^{87}\text{Sr}/^{86}\text{Sr}$ ratio (denoted as $R$) in a magma chamber. The magma, with an initial ratio $R_0$, is contaminated by wall rock with a ratio $R_c$ at a constant rate. The simplified ODE for the rate of change of $R$ is:
$$ \frac{dR}{dt} = k (R_c - R) $$
where $k$ is the effective contamination rate constant.

**Tasks:**
1.  **Analytical Solution (Optional Bonus):** Derive or write down the analytical solution $R(t)$ given $R(0) = R_0$.
2.  **Numerical Solution with Explicit Euler:** Given the parameters:
    - Initial magma ratio, $R_0 = 0.704$
    - Contaminant ratio, $R_c = 0.710$
    - Contamination rate, $k = 5 \times 10^{-4} \text{ (time units)}^{-1}$
    - Total simulation time, $T_{final} = 5000$ time units.
    Write a Python script to solve this ODE for $R(t)$ using the Explicit Euler method.
3.  **Time Step Exploration:**
    - Solve the ODE using a "reasonable" time step (e.g., $\Delta t = 100$).
    - Investigate what happens with a very large time step (e.g., $\Delta t = 2500$). (Hint: The equation can be rewritten as $\frac{dR}{dt} = -k(R - R_c)$).
4.  **Plotting:** Plot your numerical solutions for both time steps. If you found the analytical solution, include it for comparison.


```python
# E3.1: Explicit Euler for Sr Isotope Geochemistry

import numpy as np
import matplotlib.pyplot as plt

# 1. Analytical Solution (Bonus)
# The ODE is dR/dt = -k(R - Rc). The solution is R(t) = Rc + (R0 - Rc) * exp(-kt)
def analytical_sr_solution(t, R0, Rc, k):
    return Rc + (R0 - Rc) * np.exp(-k * t)

# 2. Numerical Solution with Explicit Euler
# Parameters
R0 = 0.704
Rc = 0.710
k = 5e-4
T_final = 5000.0

# --- Simulation with "reasonable" time step ---
dt1 = 100.0
time1 = np.arange(0, T_final + dt1, dt1)
R_sol1 = np.zeros_like(time1)
R_sol1[0] = R0

for i in range(len(time1) - 1):
    R_sol1[i+1] = R_sol1[i] + dt1 * k * (Rc - R_sol1[i])
    # Alternative form: R_sol1[i+1] = R_sol1[i] * (1 - k*dt1) + dt1*k*Rc

print(f"Simulation with dt = {dt1} completed.")

# --- Simulation with large (unstable) time step ---
# Stability limit check: 1 - k*dt >= -1  =>  k*dt <= 2  => dt <= 2/k = 2/5e-4 = 4000
# dt=2500 should be stable but oscillatory.
dt2 = 2500.0
time2 = np.arange(0, T_final + dt2, dt2)
R_sol2 = np.zeros_like(time2)
R_sol2[0] = R0

for i in range(len(time2) - 1):
    R_sol2[i+1] = R_sol2[i] + dt2 * k * (Rc - R_sol2[i])

print(f"Simulation with dt = {dt2} completed.")

# Analytical solution for plotting
time_analytical = np.linspace(0, T_final, 500)
R_analytical = analytical_sr_solution(time_analytical, R0, Rc, k)

# 4. Plotting
plt.figure(figsize=(10, 6))
plt.plot(time_analytical, R_analytical, 'k-', lw=2, label='Analytical Solution')
plt.plot(time1, R_sol1, 'bo--', label=f'Explicit Euler (dt={dt1})')
plt.plot(time2, R_sol2, 'ro--', label=f'Explicit Euler (dt={dt2})')

plt.xlabel("Time (years)")
plt.ylabel("$^{87}$Sr/$^{86}$Sr Ratio (R)")
plt.title("Sr Isotope Evolution in a Magma Chamber")
plt.legend()
plt.grid(True)
plt.show()
```

### E3.2: Implicit Euler for Sr Isotope Evolution

Continuing with the Sr isotope evolution problem from Exercise E3.1:

**Tasks:**

3.  **Test with Large Time Step:** Use the same large time step that caused issues for the Explicit Euler method (e.g., $\Delta t = 2500$).
4.  **Plotting and Comparison:** On a new plot, show the analytical solution and the Implicit Euler solutions for both $\Delta t = 100$ and $\Delta t = 2500$. Comment on the stability and accuracy of the Implicit Euler method compared to the Explicit Euler method.

```python
# E3.2: Implicit Euler for Sr Isotope Evolution

import numpy as np
import matplotlib.pyplot as plt

# 1. Derive Update Formula
# dR/dt = k(Rc - R)
# (R_k+1 - R_k) / dt = k(Rc - R_k+1)
# R_k+1 - R_k = dt*k*Rc - dt*k*R_k+1
# R_k+1 * (1 + k*dt) = R_k + dt*k*Rc
# R_k+1 = (R_k + dt*k*Rc) / (1 + k*dt)  <-- This is the update formula

# Parameters (same as E3.1)
R0 = 0.704
Rc = 0.710
k = 5e-4
T_final = 5000.0

# 2. Python Implementation

# --- Simulation with reasonable time step ---
dt1 = 100.0
time1_imp = np.arange(0, T_final + dt1, dt1)
R_sol1_imp = np.zeros_like(time1_imp)
R_sol1_imp[0] = R0

for i in range(len(time1_imp) - 1):
    R_sol1_imp[i+1] = (R_sol1_imp[i] + dt1 * k * Rc) / (1 + k * dt1)

print(f"Implicit simulation with dt = {dt1} completed.")

# 3. Test with Large Time Step
dt2 = 2500.0
time2_imp = np.arange(0, T_final + dt2, dt2)
R_sol2_imp = np.zeros_like(time2_imp)
R_sol2_imp[0] = R0

for i in range(len(time2_imp) - 1):
    R_sol2_imp[i+1] = (R_sol2_imp[i] + dt2 * k * Rc) / (1 + k * dt2)

print(f"Implicit simulation with dt = {dt2} completed.")

# Analytical solution for plotting (reused from E3.1)
time_analytical = np.linspace(0, T_final, 500)
R_analytical = Rc + (R0 - Rc) * np.exp(-k * time_analytical)

# 4. Plotting and Comparison
plt.figure(figsize=(10, 6))
plt.plot(time_analytical, R_analytical, 'k-', lw=2, label='Analytical Solution')
plt.plot(time1_imp, R_sol1_imp, 'go--', label=f'Implicit Euler (dt={dt1})')
plt.plot(time2_imp, R_sol2_imp, 'mo--', label=f'Implicit Euler (dt={dt2})')

plt.xlabel("Time (years)")
plt.ylabel("$^{87}$Sr/$^{86}$Sr Ratio (R)")
plt.title("Sr Isotope Evolution - Implicit Euler Method")
plt.legend()
plt.grid(True)
plt.show()

# Comment: The implicit method remains stable and non-oscillatory even with the large dt=2500,
# whereas the explicit method showed strong oscillations. However, the accuracy with dt=2500 is poor,
# as it significantly lags behind the true solution.
```

### E3.3: Bisection Method - Finding a Bracket and Root

Consider the function $f(x) = x^3 - x - 2$. We want to find a root of this function.

**Tasks:**
1.  **Plot the Function:** Write a short Python script to plot $f(x)$ for $x$ in the range, say, $[-2, 3]$.
2.  **Find a Bracketing Interval [a,b]:** By observing the plot, find an interval $[a,b]$ of width at least 0.5 such that $f(a)$ and $f(b)$ have opposite signs. State your chosen $a$ and $b$, and the values of $f(a)$ and $f(b)$.
3.  **Manual Bisection (First 3 Iterations):** Starting with your chosen interval, perform the first three iterations of the bisection method by hand (e.g., with pen and paper, not by writing code). For each iteration, calculate the midpoint $c$, $f(c)$, and state the new interval.
4.  **Python Implementation for Bracketing (Challenge):** (Optional Challenge) Describe a simple strategy for a Python function that tries to automatically find a bracketing interval for a root of a general function `func(x)`, given an initial guess and a step size.

```python
# E3.3: Bisection Method

import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**3 - x - 2

# 1. Plot the Function
x_vals = np.linspace(-2, 3, 400)
y_vals = f(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, label='$f(x) = x^3 - x - 2$')
plt.axhline(0, color='black', lw=0.5)
plt.title("Plot of the function to find its root")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()

# 2. Find a Bracketing Interval
# From the plot, the root is between 1 and 2. Let's test a=1 and b=2.
a = 1.0
b = 2.0
f_a = f(a)
f_b = f(b)
print(f"2. Bracketing Interval:")
print(f"   Chosen interval: [{a}, {b}]")
print(f"   f(a) = f({a}) = {f_a}")
print(f"   f(b) = f({b}) = {f_b}")
print(f"   Check sign change: f(a) * f(b) = {f_a * f_b} (which is < 0, so it's a valid bracket)")

# 3. Manual Bisection (First 3 Iterations)
print("\n3. Manual Bisection Steps:")
# Iteration 1
c1 = (a + b) / 2
f_c1 = f(c1)
print(f"   Iter 1: c = {c1:.4f}, f(c) = {f_c1:.4f}. New interval: [{c1:.4f}, {b:.4f}]")
a1, b1 = c1, b

# Iteration 2
c2 = (a1 + b1) / 2
f_c2 = f(c2)
print(f"   Iter 2: c = {c2:.4f}, f(c) = {f_c2:.4f}. New interval: [{a1:.4f}, {c2:.4f}]")
a2, b2 = a1, c2

# Iteration 3
c3 = (a2 + b2) / 2
f_c3 = f(c3)
print(f"   Iter 3: c = {c3:.4f}, f(c) = {f_c3:.4f}. New interval: [{a2:.4f}, {c3:.4f}]")
```

### E3.4: Nonlinear ODE - Simplified Population Logistics with Harvesting

Consider a simplified model for a fish population $P(t)$ that grows logistically but is also subject to constant rate harvesting $H$:
$$ \frac{dP}{dt} = r P \left(1 - \frac{P}{K}\right) - H $$
where $r$ is the growth rate, $K$ is the carrying capacity, and $H$ is the harvesting rate. This is a nonlinear ODE.

Let $P_0 = 500$ fish, $r = 0.2 \text{ year}^{-1}$, $K = 1000$ fish, and $H = 70$ fish/year. We want to find the population $P(t)$ over 50 years using the Implicit Euler method and the bisection method.

**Tasks:**
1.  **Formulate $G(P^{k+1})=0$:** Apply the Implicit Euler scheme to the ODE and rearrange it to get a nonlinear algebraic equation of the form $G(P^{k+1}) = 0$. Write down your function $G(X)$ where $X=P^{k+1}$.
2.  **Python Implementation:** Write a Python script to solve this problem from $t=0$ to $t=50$ years with a time step $\Delta t = 1$ year.
    - Define a Python function for $G(X)$.
    - In your time-stepping loop, use an inlined bisection method to find the root $P^{k+1}$ at each step. Determine a suitable bracketing interval (e.g., $[0, K*1.5]$).
3.  **Plotting:** Plot the fish population $P(t)$ versus time.
4.  **Discussion (Conceptual):** What might happen if the harvesting rate $H$ is very high (e.g., $H=150$ fish/year)? How would this affect your choice of bracketing interval or the existence of a positive solution?

```python
# E3.4: Nonlinear ODE with Bisection Method

import numpy as np
import matplotlib.pyplot as plt

# Parameters
P0 = 500      # fish
r = 0.2       # 1/year
K = 1000      # fish
H = 70        # fish/year
T_final = 50  # years
dt = 1.0      # year

# 1. Formulate G(P_n+1) = 0
# P_n+1 - P_n = dt * [r*P_n+1*(1 - P_n+1/K) - H]
# (r/K)*P_n+1^2 + (1-r)*P_n+1 - (P_n - dt*H) = 0  --> This is the quadratic form.
# Let's use the rearranged form directly for G(X)
def G_population(X, P_prev, dt, r, K, H):
    """Function G(X)=0 for the implicit population model."""
    return X - P_prev - dt * (r * X * (1 - X / K) - H)

# 2. Python Implementation
time = np.arange(0, T_final + dt, dt)
P_sol = np.zeros_like(time)
P_sol[0] = P0

tol_bisect = 1e-6
max_iter = 100

for k in range(len(time) - 1):
    P_k = P_sol[k]

    # Determine bracketing interval [a, b]
    a = 0.0
    b = K * 1.5 # A reasonable upper bound, well above carrying capacity

    Ga = G_population(a, P_k, dt, r, K, H)
    Gb = G_population(b, P_k, dt, r, K, H)

    if Ga * Gb >= 0:
        print(f"Error: Could not bracket root at time step {k}")
        break

    # Bisection loop
    for i in range(max_iter):
        c = (a + b) / 2
        Gc = G_population(c, P_k, dt, r, K, H)

        if abs(Gc) < tol_bisect:
            break

        if Ga * Gc < 0:
            b = c
        else:
            a = c
            Ga = Gc

    P_sol[k+1] = c

# 3. Plotting
plt.figure(figsize=(10, 6))
plt.plot(time, P_sol, 'o-', label='Fish Population')
plt.axhline(K, color='gray', linestyle='--', label='Carrying Capacity (K)')
plt.xlabel("Time (years)")
plt.ylabel("Population (number of fish)")
plt.title("Logistic Growth with Constant Harvesting (Implicit Euler)")
plt.legend()
plt.grid(True)
plt.show()

# 4. Discussion
# If H is very high (e.g., H=150), it might exceed the maximum sustainable yield.
# This could lead to a situation where G(X) is always positive for X>0, meaning
# there is no positive root, and the bisection method would fail to find a bracket.
# Physically, this means the population would decline towards extinction.
```
